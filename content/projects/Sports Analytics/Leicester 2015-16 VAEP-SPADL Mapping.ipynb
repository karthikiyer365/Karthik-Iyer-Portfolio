# Storyline

## Context

Leicester City won the 2015/16 Premier League with 81 points, 10 clear, yet ranked 18th in touches. Counting possession cannot explain that, so the question was what each on-ball action was worth. See the [write-up](https://writing.karthikiyer.info/leicester-2015-16-vaep/) and the interactive [Team Analysis page](https://sports.karthikiyer.info/team?lg=eng-premier-league-2015&team=22), which shows where a team creates value across 30 pitch zones.

## Workflow

StatsBomb events for 380 matches → SPADL action table → game-state features and score / concede labels → two XGBoost classifiers → per-action value → team, player, and zone aggregates → validation and calibration checks.

## Technical Implementation

- **SPADL:** each action becomes one row with its type, player, start and end location on a 105 × 68 m pitch, and whether it worked (758,434 actions).
- **Features:** ball location, action type, result, and the two preceding actions.
- **Labels:** did the team score within the next 10 actions, and did it concede within the next 10.
- **Models:** two XGBoost classifiers, one for each label. Action value is the change in scoring chance minus the change in conceding chance; a typical action is worth about ±0.01.
- **Validation:** games split by date into two halves; a model trained on one half rates the other, then they swap, so no action is valued by a model that saw its game.

## Outcomes

Leicester ranked 1st for value per touch (0.234 vs 0.162 league average) and 18th for touches, with 52% of value coming from the final third and shots making up 1.6% of actions against 1.2% league-wide. Mahrez (+17.1), Vardy (+13.3), and Drinkwater (+8.5) produced 53% of team value. The model places Leicester 6th overall, below Arsenal, and 11th for defensive value despite the 4th-best goals-conceded record. Calibration held within 0.5 percentage points and team value correlated 0.768 with the table, but player rankings were weakly reliable across season halves (0.313). Event data is on-ball only, so off-ball play and defensive positioning stay invisible.
